# Example use of smetplots



```python
import numpy as np
import matplotlib.pyplot as plt
from snowpacktools.smetplots import smet_plot 
# importlib.reload(smet_plot)
```

```python

# path to the smet files:
path = "/home/alkr/model_out/snp-pixel-bjoberg-2024-25"

# name of the smet files:
file1500= "VIR3A.smet"
file900= "VIR1A.smet"




# read the smet file (returns the data as a pandas dataframe, and header as a dictionary):
smet1500,  header1500 = smet_plot.SMET2df(f"{path}/{file1500}", keyword='[DATA]', num_lines=None)

smet900, header900 = smet_plot.SMET2df(f"{path}/{file900}", keyword='[DATA]', num_lines=None)

print(header1500)
smet1500.head() 

```

## Plot a timeseries for several parameters

```python
vars_to_plot = ['dIntEnergySnow', 'meltFreezeEnergySnow', 'ColdContentSnow','LWR_net'] # ,  'ISWR_diff', 'TA', 'TSS_mod'	
# Define the datetime range
start_date = '2025-01-05'
end_date = '2025-01-21'
```

```python
smet_plot.plot_vars(smet900, vars_to_plot, start_date, end_date, header900) # title is automatically set to the station name, lat, lon, and elevation.
```

## Plot multiple timeseries (stations) in one plot by defining a fig and axes first:

```python

# make a fig and axes, so we can plot several timeseries on the same figure:
fig, ax = plt.subplots( int(np.ceil(len(vars_to_plot)/2)),2,figsize=(7*2, 4*np.ceil(len(vars_to_plot)/2)))

smet_plot.plot_vars(smet900, vars_to_plot, start_date, end_date, header900, fig=fig, ax=ax, label='900 m')
smet_plot.plot_vars(smet1500, vars_to_plot, start_date, end_date, header1500, fig=fig, ax=ax, label='1500 m', title="Virtual stations at 900 m and 1500 m")
```
